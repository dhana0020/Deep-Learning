Implementation of Logic Function XOR using Feed forward network


A feed-forward neural network with one hidden layer is trained using the backpropagation algorithm to learn the XOR logic function. The hidden layer introduces non-linearity, enabling the network to correctly classify XOR inputs.

```python
import numpy as np
import random
```

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))

def der_sigmoid(x):
    return x*(1-x)
```

```python

X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

Y = np.array([[0],[1],[1],[0]])


hidden_weights = np.random.uniform(-0.5, 0.5, (2,2))
hidden_bias = np.random.uniform(-1, 1, (1,2))

output_weights = np.random.uniform(-0.5, 0.5, (2,1))
output_bias = np.random.uniform(-1, 1, (1,1))

print("Initial weights : ")
print(*hidden_weights)
print(*output_weights)
print("Initial Bias : ")
print(*hidden_bias)
print(*output_bias)
lr = 0.3
epochs = 10000


for _ in range(epochs):

    
    hidden_input = np.dot(X, hidden_weights) + hidden_bias
    hidden_output = sigmoid(hidden_input)

    output_input = np.dot(hidden_output, output_weights) + output_bias
    y_pred = sigmoid(output_input)

    
    error = Y - y_pred
    delta_out = error * der_sigmoid(y_pred)


    delta_hidden = delta_out.dot(output_weights.T) * der_sigmoid(hidden_output)

    
    output_weights += lr * hidden_output.T.dot(delta_out)
    output_bias += lr * np.sum(delta_out, axis=0, keepdims=True)

    hidden_weights += lr * X.T.dot(delta_hidden)
    hidden_bias += lr * np.sum(delta_hidden, axis=0, keepdims=True)


print("Final weights : ")
print(*hidden_weights)
print(*output_weights)
print("Final Bias : ")
print(*hidden_bias)
print(*output_bias)



print("XOR Results:")
for x, y in zip(X, y_pred):
    print(x[0], x[1], "->", 1 if y >= 0.5 else 0)
```
