```python
import tensorflow as tf
import tensorflow_datasets as tfds
from tensorflow.keras.layers import *
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam
```

```python
(ds_train, ds_test), ds_info = tfds.load(
    'cats_vs_dogs',
    split=['train[:80%]', 'train[80%:]'],
    as_supervised=True,
    with_info=True
)
```

```python
IMG_SIZE = (227, 227)
BATCH_SIZE = 32
def preprocess(image, label):
    image = tf.image.resize(image, IMG_SIZE)
    image = image / 255.0
    return image, label
```

```python
ds_train = ds_train.map(preprocess)\
                   .batch(BATCH_SIZE)\
                   .prefetch(tf.data.AUTOTUNE)

ds_test = ds_test.map(preprocess)\
                 .batch(BATCH_SIZE)\
                 .prefetch(tf.data.AUTOTUNE)
```

```python
def AlexNet():
    inputs = Input(shape=(227,227,3))

    x = Conv2D(96, (11,11), strides=4, activation='relu')(inputs)
    x = MaxPooling2D((3,3), strides=2)(x)

    x = Conv2D(256, (5,5), padding='same', activation='relu')(x)
    x = MaxPooling2D((3,3), strides=2)(x)

    x = Conv2D(384, (3,3), padding='same', activation='relu')(x)
    x = Conv2D(384, (3,3), padding='same', activation='relu')(x)
    x = Conv2D(256, (3,3), padding='same', activation='relu')(x)
    x = MaxPooling2D((3,3), strides=2)(x)

    x = Flatten()(x)
    x = Dense(4096, activation='relu')(x)
    x = Dropout(0.5)(x)
    x = Dense(4096, activation='relu')(x)
    x = Dropout(0.5)(x)

    outputs = Dense(1, activation='sigmoid')(x)

    return Model(inputs, outputs)
```

```python
def ZFNet():
    inputs = Input(shape=(227,227,3))

    x = Conv2D(96, (7,7), strides=2, activation='relu')(inputs)
    x = MaxPooling2D((3,3), strides=2)(x)

    x = Conv2D(256, (5,5), padding='same', activation='relu')(x)
    x = MaxPooling2D((3,3), strides=2)(x)

    x = Conv2D(384, (3,3), padding='same', activation='relu')(x)
    x = Conv2D(384, (3,3), padding='same', activation='relu')(x)
    x = Conv2D(256, (3,3), padding='same', activation='relu')(x)
    x = MaxPooling2D((3,3), strides=2)(x)

    x = Flatten()(x)
    x = Dense(4096, activation='relu')(x)
    x = Dropout(0.5)(x)
    x = Dense(4096, activation='relu')(x)
    x = Dropout(0.5)(x)

    outputs = Dense(1, activation='sigmoid')(x)

    return Model(inputs, outputs)
```

```python
alexnet = AlexNet()
zfnet = ZFNet()
```

```python
alexnet.compile(
    optimizer=Adam(0.0001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

zfnet.compile(
    optimizer=Adam(0.0001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
print("ALEXNET SUMMARY")
alexnet.summary()

print("\nZFNET SUMMARY")
zfnet.summary()
```

```python
alexnet.fit(
    ds_train,
    epochs=10,
    validation_data=ds_test
)
zfnet.fit(
    ds_train,
    epochs=10,
    validation_data=ds_test
)
alex_loss, alex_acc = alexnet.evaluate(ds_test, verbose=0)
zf_loss, zf_acc = zfnet.evaluate(ds_test, verbose=0)
```

```python

print("\nMODEL PERFORMANCE COMPARISON")
print("--------------------------------")
print(f"AlexNet  - Accuracy: {alex_acc*100:.2f}% | Loss: {alex_loss:.4f}")
print(f"ZFNet    - Accuracy: {zf_acc*100:.2f}% | Loss: {zf_loss:.4f}")
```
